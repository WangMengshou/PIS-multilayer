```python
y1end_10_10_5 = mcmc_features_times_10_10_5[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[0:Tscale+1][-1]
y2end_10_10_5 = mcmc_features_times_con_10_10_5[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[-1]

y1end_10_X_5 = mcmc_features_times_10_X_5[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[0:Tscale+1][-1]
y2end_10_X_5 = mcmc_features_times_con_10_X_5[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[-1]

y1end_10_10_X = mcmc_features_times_10_10_X[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[0:Tscale+1][-1]
y2end_10_10_X = mcmc_features_times_con_10_10_X[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[-1]

y1end_10_X_X = mcmc_features_times_10_X_X[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[0:Tscale+1][-1]
y2end_10_X_X = mcmc_features_times_con_10_X_X[..., [2, 3, 4]].sum(dim=2).mean(dim=1)[-1]

# Creating bar chart for final epidemic sizes
fig, ax = plt.subplots(figsize=(12, 5))

# Data for the bar chart
scenarios = ['Panel (a)', 'Panel (b)', 'Panel (c)', 'Panel (d)']
feedback_values = [y1end_10_10_5.item(), y1end_10_X_5.item(), y1end_10_10_X.item(), y1end_10_X_X.item()]
constant_values = [y2end_10_10_5.item(), y2end_10_X_5.item(), y2end_10_10_X.item(), y2end_10_X_X.item()]

# Set the width of the bars
bar_width = 0.4
import numpy as np
x = np.arange(len(scenarios))

# Create the bars
bars1 = ax.bar(x - bar_width/2, feedback_values, bar_width, label='Feedback', color='#1f77b4', alpha=0.95)
bars2 = ax.bar(x + bar_width/2, constant_values, bar_width, label='Constant', color='#ff7f0e', alpha=0.95)

# Add labels, title and legend
ax.set_ylabel('Steady-state epidemic size', fontsize=20)
ax.set_xticks(x)
ax.set_xticklabels(scenarios, fontsize=20)
ax.tick_params(axis='y', labelsize=20)
ax.legend(fontsize=20)
ax.set_ylim(0.3, 0.75)

# Add value labels on top of bars
def add_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=20)

add_labels(bars1)
add_labels(bars2)

plt.tight_layout()
plt.savefig('../figs/Two_layer/final_epidemic_comparison.svg', bbox_inches='tight')
plt.savefig('../figs/Two_layer/final_epidemic_comparison.png', dpi=600, bbox_inches='tight')
plt.show()
```
